# Bài tập Hồi quy tuyến tính

File `Salary_Data.csv` nằm cùng thư mục với notebook trong `Practice/`.

Mỗi dòng là dữ liệu của một người:
- `YearsExperience`: số năm kinh nghiệm, dùng làm feature (đầu vào).
- `Salary`: mức lương, dùng làm label (giá trị cần dự đoán).

Mục tiêu là huấn luyện mô hình để dự đoán mức lương dựa trên số năm kinh nghiệm.

## Bước 1: Tải dữ liệu

Dùng `pandas` để đọc file CSV vào biến `data`, sau đó xem 5 dòng đầu tiên để kiểm tra dữ liệu. Code sẽ tìm file trong thư mục hiện tại hoặc trong `Practice/`.

```python
from pathlib import Path
import pandas as pd

# Tìm file CSV ở thư mục đang chạy notebook hoặc trong thư mục Practice.
csv_path = Path("Salary_Data.csv")
if not csv_path.exists():
    csv_path = Path("Practice") / "Salary_Data.csv"

# Đọc dữ liệu thành bảng và xem thử 5 dòng đầu tiên.
data = pd.read_csv(csv_path)
data.head()
```

## Bước 2: Tách feature và label

- `features` chứa cột `YearsExperience`, là thông tin đầu vào của mô hình. Dùng hai cặp ngoặc vuông để giữ dữ liệu dạng bảng 2 chiều.
- `label` chứa cột `Salary`, là kết quả mô hình cần học cách dự đoán.

```python
# Feature là số năm kinh nghiệm; dùng hai cặp ngoặc để giữ dữ liệu dạng bảng 2 chiều.
features = data[["YearsExperience"]]

# Label là mức lương cần dự đoán.
label = data["Salary"]
```

## Bước 3: Huấn luyện mô hình

Tạo `LinearRegression` và huấn luyện bằng `features` cùng `label`.

- `weight` (hệ số góc) cho biết lương dự đoán thay đổi bao nhiêu khi kinh nghiệm tăng thêm 1 năm.
- `bias` (hệ số chặn) là lương dự đoán khi kinh nghiệm bằng 0 năm.

Công thức mô hình: `Salary = bias + weight * YearsExperience`.

```python
from sklearn.linear_model import LinearRegression

# Tạo mô hình và cho mô hình học mối quan hệ giữa kinh nghiệm và lương.
model = LinearRegression()
model.fit(features, label)

# Hệ số góc cho biết lương thay đổi bao nhiêu khi kinh nghiệm tăng 1 năm.
weight = model.coef_[0]

# Hệ số chặn là mức lương mô hình ước tính khi kinh nghiệm bằng 0 năm.
bias = model.intercept_

print(f"Hệ số góc (weight): {weight:.2f}")
print(f"Hệ số chặn (bias): {bias:.2f}")
print(f"Công thức: Salary = {bias:.2f} + {weight:.2f} * YearsExperience")
```

## Bước 4: Dự đoán

Tạo bảng đầu vào cho người có 5 năm kinh nghiệm rồi gọi `model.predict`. Dữ liệu đưa vào `predict` cần có dạng bảng 2 chiều. Kết quả dự kiến khoảng `73,000`.

```python
# Tạo một bảng 2 chiều cho người có 5 năm kinh nghiệm.
years = pd.DataFrame({"YearsExperience": [5]})

# Dùng mô hình đã huấn luyện để dự đoán lương; lấy kết quả duy nhất đầu tiên.
predicted_salary = model.predict(years)[0]
print(f"Lương dự đoán cho 5 năm kinh nghiệm: {predicted_salary:,.2f}")
```
